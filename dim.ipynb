```python
import numpy as np
import pandas as pd
from sklearn.manifold import TSNE
```

```python
df = pd.read_parquet('processed/combined.parquet')
```

```python
df
```

```python
count = 0 
for index, row in df.iterrows():
    if row['best_match_index'] == row['predicted_missing_id']:
        print(index)
        count += 1
```

```python
count/len(df)
```

```python
skills = np.vstack(df['skill_embeddings_ordered'].values)
jobs   = np.vstack(df['best_match_job_embedding'].values)
diffs  = np.vstack(df['difference'])
predit = np.vstack(df['predicted_missing_embedding'].values)
```

```python
X = np.vstack([skills, jobs, diffs, predit])
```

```python
tsne = TSNE(n_components=3, perplexity=40, max_iter=2000, learning_rate='auto')
```

```python
X_2d = tsne.fit_transform(X)
```

```python
n = len(df)

skills_3d = X_2d[0:n]
jobs_3d   = X_2d[n:2*n]
diff_3d =   X_2d[2*n:3*n]
predit_3d = X_2d[3*n:4*n]
```

```python
df["skill_tsne"]   = list(skills_3d)
df["job_tsne"]     = list(jobs_3d)
df['diff_tsne'] =    list(diff_3d)
df["predict_tsne"]  = list(predit_3d)
```

```python
df
```

```python
df.to_parquet('processed/dim_reduced.parquet')
```
